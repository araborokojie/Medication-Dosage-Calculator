# Task 16- Generate a Medication Dosage Report

```python
hospital_name = "SmartCare Hospital"

# Patient Information
patient_name = "Daniel Okafor"
patient_id = "PT001"
age = 35
weight = 70
medication = "Med-X"
dosage_rate = 10
doses_per_day = 2
maximum_daily_dose = 1500


# Calculate Single Dose
def calculate_dose(weight, dosage_rate):
    single_dose = weight * dosage_rate
    return single_dose


# Calculate Total Daily Dose
def calculate_daily_dose(single_dose, doses_per_day):
    total_daily_dose = single_dose * doses_per_day
    return total_daily_dose


# Validate Patient Information
def validate_patient(age, weight, patient_id):
    age_valid = age > 0 and age <= 120
    weight_valid = weight > 0 and weight <= 300
    patient_id_valid = patient_id != ""
    return age_valid, weight_valid, patient_id_valid


# Calculate Age Category
if age <= 12:
    patient_category = "Child"
elif age <= 17:
    patient_category = "Teenager"
elif age <= 64:
    patient_category = "Adult"
else:
    patient_category = "Senior"


# Determine Dosage Frequency
if doses_per_day == 1:
    dosage_frequency = "Once Daily"
elif doses_per_day == 2:
    dosage_frequency = "Twice Daily"
elif doses_per_day == 3:
    dosage_frequency = "Three Times Daily"
else:
    dosage_frequency = "Custom Frequency"


# Perform Calculations
single_dose = calculate_dose(weight, dosage_rate)
total_daily_dose = calculate_daily_dose(single_dose, doses_per_day)


# Perform Validation
age_valid, weight_valid, patient_id_valid = validate_patient(age, weight, patient_id)


# Check Daily Dose Limit
if total_daily_dose > maximum_daily_dose:
    daily_dose_status = "Requires Review"
else:
    daily_dose_status = "Within Defined Limit"


# Determine Overall Patient Record Status
if age_valid and weight_valid and patient_id_valid:
    patient_record_status = "Valid"
else:
    patient_record_status = "Invalid"


# Generate Report
print("========================================")
print("       MEDICATION DOSAGE REPORT")
print("========================================")
print("Patient Name:", patient_name)
print("Patient ID:", patient_id)
print("Age:", age)
print("Age Category:", patient_category)
print("Weight:", weight, "kg")
print("Medication:", medication)
print("Dosage Rate:", dosage_rate, "mg/kg")
print("Doses Per Day:", doses_per_day)
print("========================================")
print("")
print("")

print("========================================")
print("       DOSAGE CALCULATION")
print("========================================")
print("Single Dose:", single_dose, "mg")
print("Total Daily Dose:", total_daily_dose, "mg")
print("Maximum Daily Dose:", maximum_daily_dose, "mg")
print("Dosage Frequency:", dosage_frequency)
print("========================================")
print("")
print("")

print("========================================")
print("             VALIDATION")
print("========================================")
print("Age Valid:", age_valid)
print("Weight Valid:", weight_valid)
print("Patient ID Valid:", patient_id_valid)
print("Daily Dose Status:", daily_dose_status)
print("Patient Record:", patient_record_status)
print("========================================")
```

# BONUS CHALLENGE

```python
# PATIENT DOSAGE PROCESSING SYSTEM

hospital_name = "SmartCare Hospital"
maximum_daily_dose = 1500

# Fictional patient records
patients = [
    {"name": "Daniel Okafor", "patient_id": "PT001", "age": 35, "weight": 70, "dosage_rate": 10, "doses_per_day": 2},
    {"name": "Grace Bello", "patient_id": "PT002", "age": 68, "weight": 60, "dosage_rate": 8, "doses_per_day": 2},
    {"name": "Samuel Ade", "patient_id": "PT003", "age": 15, "weight": 50, "dosage_rate": 6,"doses_per_day": 3},
    {"name": "Anita James", "patient_id": "PT004", "age": 130, "weight": 65, "dosage_rate": 10, "doses_per_day": 2}
]


# Function to calculate single dose
def calculate_dose(weight, dosage_rate):
    single_dose = weight * dosage_rate
    return single_dose


# Function to calculate total daily dose
def calculate_daily_dose(single_dose, doses_per_day):
    total_daily_dose = single_dose * doses_per_day
    return total_daily_dose


# Function to validate patient
def validate_patient(age, weight, patient_id):
    age_valid = age > 0 and age <= 120
    weight_valid = weight > 0 and weight <= 300
    patient_id_valid = patient_id != ""
    return age_valid, weight_valid, patient_id_valid


# Counters
passed_validation = 0
failed_validation = 0

# Variables for highest doses
highest_single_dose = 0
highest_single_dose_patient = ""

highest_daily_dose = 0
highest_daily_dose_patient = ""

# Lists for special categories
senior_patients = []
invalid_age_patients = []
review_patients = []


# Process all patients using a for loop
for patient in patients:

    name = patient["name"]
    patient_id = patient["patient_id"]
    age = patient["age"]
    weight = patient["weight"]
    dosage_rate = patient["dosage_rate"]
    doses_per_day = patient["doses_per_day"]

    # Calculate doses
    single_dose = calculate_dose(weight, dosage_rate)
    total_daily_dose = calculate_daily_dose(single_dose, doses_per_day)

    # Validate patient
    age_valid, weight_valid, patient_id_valid = validate_patient(age, weight, patient_id)

    # Check overall validation
    if age_valid and weight_valid and patient_id_valid:
        passed_validation += 1
        validation_status = "Valid"
    else:
        failed_validation += 1
        validation_status = "Invalid"

    # Determine age category
    if age <= 12:
        age_category = "Child"
    elif age <= 17:
        age_category = "Teenager"
    elif age <= 64:
        age_category = "Adult"
    else:
        age_category = "Senior"

    # Find senior patients
    if age_category == "Senior":
        senior_patients.append(name)

    # Find patients with invalid age
    if not age_valid:
        invalid_age_patients.append(name)

    # Check maximum daily dose
    if total_daily_dose > maximum_daily_dose:
        review_patients.append(name)

    # Find highest single dose
    if single_dose > highest_single_dose:
        highest_single_dose = single_dose
        highest_single_dose_patient = name

    # Find highest daily dose
    if total_daily_dose > highest_daily_dose:
        highest_daily_dose = total_daily_dose
        highest_daily_dose_patient = name

    # Display individual patient results
    print("----------------------------------------")
    print("Patient:", name)
    print("Single Dose:", single_dose, "mg")
    print("Total Daily Dose:", total_daily_dose, "mg")
    print("Age Category:", age_category)
    print("Doses Per Day:", doses_per_day)
    print("Validation Status:", validation_status)

    if total_daily_dose > maximum_daily_dose:
        print("Daily Dose Status: Requires Review")
    else:
        print("Daily Dose Status: Within Defined Limit")
        print("")



# FINAL BONUS SUMMARY

print("========================================")
print("           BONUS CHALLENGE")
print("========================================")

print("Patient Records Passed Validation:", passed_validation)

print("Patient Records Failed Validation:", failed_validation)

print("Highest Calculated Single Dose:", highest_single_dose, "mg")

print("Patient:", highest_single_dose_patient)

print("Highest Calculated Daily Dose:", highest_daily_dose, "mg")

print("Patient:", highest_daily_dose_patient)

print("Senior Patient(s):", ", ".join(senior_patients))

print("Patient(s) With Invalid Age:", ", ".join(invalid_age_patients))

print("Patient(s) Requiring Daily Dose Review:", ", ".join(review_patients))

print("\nDoses Per Day for Each Patient:")

for patient in patients:
    print(patient["name"], ":", patient["doses_per_day"])

print("========================================")
```
